## Limpeza inicial

A coluna `Income` possui 24 valores ausentes. Como essa quantidade representa uma parte pequena da base, optei por remover essas linhas para não distorcer análises relacionadas à renda e consumo.

```python
df_clean = df.dropna(subset=["Income"]).copy()

df_clean.shape
```

```python
df_clean["Income"].isna().sum()
```

```python
df_clean["Dt_Customer"] = pd.to_datetime(df_clean["Dt_Customer"], dayfirst=True)

df_clean[["Dt_Customer"]].head()
```

```python
df_clean["Age"] = 2026 - df_clean["Year_Birth"]

df_clean["Total_Children"] = df_clean["Kidhome"] + df_clean["Teenhome"]

df_clean["Total_Spent"] = (
    df_clean["MntWines"]
    + df_clean["MntFruits"]
    + df_clean["MntMeatProducts"]
    + df_clean["MntFishProducts"]
    + df_clean["MntSweetProducts"]
    + df_clean["MntGoldProds"]
)

df_clean["Total_Purchases"] = (
    df_clean["NumWebPurchases"]
    + df_clean["NumCatalogPurchases"]
    + df_clean["NumStorePurchases"]
)

df_clean[["Age", "Income", "Total_Children", "Total_Spent", "Total_Purchases"]].head()
```

## Checagem das variáveis criadas

Depois de criar as variáveis de idade, total de filhos, gasto total e total de compras, vou fazer algumas checagens simples para entender se existem valores fora do esperado.

```python
df_clean[["Age", "Income", "Total_Spent", "Total_Purchases"]].describe()
```

```python
df_clean.sort_values("Age", ascending=False)[["ID", "Year_Birth", "Age", "Income", "Total_Spent"]].head(10)
```

```python
df_clean[df_clean["Age"] > 100][["ID", "Year_Birth", "Age", "Income", "Total_Spent"]]
```

```python
df_clean = df_clean[df_clean["Age"] <= 100].copy()

df_clean.shape
```

## Salvando a base tratada

Após a limpeza inicial e criação das variáveis principais, vou salvar uma versão tratada da base para facilitar análises futuras.

```python
df_clean.to_csv("../data/processed/customers_clean.csv", index=False)

print("Arquivo salvo com sucesso.")
```

## Análise exploratória

Nesta etapa, vou analisar o perfil dos clientes, distribuição de renda, idade, gasto total e volume de compras. A ideia é entender padrões gerais antes de partir para comparações entre grupos.

```python
df_clean[["Age", "Income", "Total_Spent", "Total_Purchases", "Total_Children"]].describe().round(2)
```

```python
df_clean["Education"].value_counts()
```

```python
education_summary = (
    df_clean
    .groupby("Education")
    .agg(
        customers=("ID", "count"),
        avg_income=("Income", "mean"),
        avg_spent=("Total_Spent", "mean"),
        avg_purchases=("Total_Purchases", "mean")
    )
    .sort_values("avg_spent", ascending=False)
    .round(2)
)

education_summary
```

```python
df_clean["Marital_Status"].value_counts()
```

```python
marital_summary = (
    df_clean
    .groupby("Marital_Status")
    .agg(
        customers=("ID", "count"),
        avg_income=("Income", "mean"),
        avg_spent=("Total_Spent", "mean"),
        avg_purchases=("Total_Purchases", "mean")
    )
    .sort_values("avg_spent", ascending=False)
    .round(2)
)

marital_summary
```

```python
df_clean[["Age", "Income", "Total_Children", "Total_Spent", "Total_Purchases"]].corr().round(2)
```

### Leitura da correlação

A renda (`Income`) tem correlação positiva com o gasto total (`Total_Spent`) e com o total de compras (`Total_Purchases`). Isso indica que clientes com maior renda tendem a gastar mais e também realizar mais compras.

O total de filhos em casa (`Total_Children`) tem correlação negativa com o gasto total. Uma possível leitura é que clientes com filhos podem ter menor gasto nas categorias analisadas, ou um comportamento de compra mais seletivo.

Também existe uma correlação forte entre `Total_Spent` e `Total_Purchases`, o que é esperado: clientes que compram mais vezes tendem a acumular maior gasto total.

### Primeiras observações

Até aqui, a análise mostra uma base de clientes com diferentes perfis de renda, idade, escolaridade e comportamento de compra.

As variáveis criadas ajudam a resumir melhor o comportamento dos clientes:
- `Total_Spent`: soma dos gastos nas categorias de produto.
- `Total_Purchases`: soma das compras feitas pelos canais web, catálogo e loja.
- `Total_Children`: total de crianças/adolescentes no domicílio.

A partir daqui, vou usar gráficos para visualizar melhor essas diferenças.

```python
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df_clean["Income"], bins=40, kde=True)
plt.title("Distribuição de renda dos clientes")
plt.xlabel("Renda")
plt.ylabel("Quantidade de clientes")
plt.tight_layout()
plt.savefig("../outputs/charts/income_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df_clean["Total_Spent"], bins=40, kde=True)
plt.title("Distribuição do gasto total dos clientes")
plt.xlabel("Gasto total")
plt.ylabel("Quantidade de clientes")
plt.tight_layout()
plt.savefig("../outputs/charts/total_spent_distribution.png")
plt.show()
```

```python
education_spent = (
    df_clean
    .groupby("Education", as_index=False)
    .agg(avg_spent=("Total_Spent", "mean"))
    .sort_values("avg_spent", ascending=False)
)

plt.figure(figsize=(10, 5))
sns.barplot(data=education_spent, x="Education", y="avg_spent")
plt.title("Gasto médio por escolaridade")
plt.xlabel("Escolaridade")
plt.ylabel("Gasto médio")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig("../outputs/charts/avg_spent_by_education.png")
plt.show()
```

```python
children_spent = (
    df_clean
    .groupby("Total_Children", as_index=False)
    .agg(
        customers=("ID", "count"),
        avg_spent=("Total_Spent", "mean")
    )
)

plt.figure(figsize=(10, 5))
sns.barplot(data=children_spent, x="Total_Children", y="avg_spent")
plt.title("Gasto médio por quantidade de filhos em casa")
plt.xlabel("Total de filhos em casa")
plt.ylabel("Gasto médio")
plt.tight_layout()
plt.savefig("../outputs/charts/avg_spent_by_children.png")
plt.show()
```

## Observações dos gráficos

A distribuição de renda mostra que a maior parte dos clientes está concentrada em faixas intermediárias, com alguns valores bem mais altos. Isso é importante porque clientes de renda muito alta podem puxar a média para cima.

A distribuição de gasto total também mostra uma concentração grande de clientes com baixo ou médio gasto, enquanto uma parcela menor apresenta consumo mais alto.

No gráfico de renda versus gasto total, aparece uma relação positiva: clientes com maior renda tendem a gastar mais. Isso confirma o resultado visto na correlação.

Na análise por escolaridade, alguns grupos apresentam gasto médio maior do que outros, o que pode indicar diferenças de comportamento de consumo entre perfis educacionais.

Na análise por quantidade de filhos em casa, o gasto médio tende a ser menor conforme aumenta o número de filhos, o que conversa com a correlação negativa observada entre `Total_Children` e `Total_Spent`.

## Segmentação simples de clientes

Para facilitar a análise, criei uma segmentação simples com base no gasto total dos clientes.

A ideia não é montar um modelo avançado, mas separar a base em três grupos para comparar perfil, renda, quantidade de compras e comportamento de consumo.

```python
df_clean["Spending_Segment"] = pd.qcut(
    df_clean["Total_Spent"],
    q=3,
    labels=["Baixo gasto", "Médio gasto", "Alto gasto"]
)

df_clean["Spending_Segment"].value_counts()
```

```python
segment_summary = (
    df_clean
    .groupby("Spending_Segment", observed=False)
    .agg(
        customers=("ID", "count"),
        avg_income=("Income", "mean"),
        avg_age=("Age", "mean"),
        avg_children=("Total_Children", "mean"),
        avg_spent=("Total_Spent", "mean"),
        avg_purchases=("Total_Purchases", "mean")
    )
    .round(2)
)

segment_summary
```

```python
segment_plot = segment_summary.reset_index()

plt.figure(figsize=(9, 5))
sns.barplot(data=segment_plot, x="Spending_Segment", y="avg_spent")
plt.title("Gasto médio por segmento de cliente")
plt.xlabel("Segmento")
plt.ylabel("Gasto médio")
plt.tight_layout()
plt.savefig("../outputs/charts/avg_spent_by_segment.png")
plt.show()
```

```python
segment_summary.to_csv("../outputs/tables/segment_summary.csv")
df_clean.to_csv("../data/processed/customers_clean.csv", index=False)

print("Base tratada e resumo por segmento salvos.")
```

### Leitura dos segmentos

A segmentação mostra diferenças claras entre os grupos de clientes.

O grupo de alto gasto apresenta maior gasto médio e tende a ter renda média superior. Esse grupo pode ser mais relevante para ações de fidelização, ofertas premium ou campanhas de relacionamento.

O grupo de baixo gasto tem menor gasto médio e pode representar clientes com menor frequência de compra, menor renda média ou menor aderência às categorias analisadas.

Essa segmentação simples ajuda a transformar a análise em uma leitura mais prática para marketing, CRM e retenção.

### Leitura dos segmentos

A segmentação mostra uma diferença clara entre os grupos de clientes.

O grupo de alto gasto tem renda média mais alta, menor média de filhos em casa e maior volume médio de compras. Isso sugere um perfil com maior potencial para ações de relacionamento, ofertas premium e campanhas de retenção.

O grupo de baixo gasto apresenta renda média menor, maior média de filhos em casa e menor número médio de compras. Esse grupo pode exigir uma abordagem diferente, com campanhas mais direcionadas, ofertas de entrada ou ações para aumentar frequência de compra.

O grupo de médio gasto fica em uma posição intermediária e pode ser um público interessante para ações de evolução de consumo, tentando aproximar parte desses clientes do comportamento do grupo de alto gasto.

```python
segment_plot = segment_summary.reset_index()

plt.figure(figsize=(9, 5))
sns.barplot(data=segment_plot, x="Spending_Segment", y="avg_spent")
plt.title("Gasto médio por segmento de cliente")
plt.xlabel("Segmento")
plt.ylabel("Gasto médio")
plt.tight_layout()
plt.savefig("../outputs/charts/avg_spent_by_segment.png")
plt.show()
```

```python
segment_summary.to_csv("../outputs/tables/segment_summary.csv")
df_clean.to_csv("../data/processed/customers_clean.csv", index=False)

print("Base tratada e resumo por segmento salvos.")
```

## Conclusões

A análise mostrou que renda, quantidade de compras e gasto total estão bem relacionados. Clientes com maior renda tendem a gastar mais e comprar com mais frequência.

Também foi possível observar que clientes com mais filhos em casa apresentam, em média, menor gasto total nas categorias analisadas. Isso pode indicar diferenças de prioridade de consumo ou menor aderência às categorias da base.

A segmentação por gasto ajudou a separar a base em três grupos simples:
- clientes de baixo gasto;
- clientes de médio gasto;
- clientes de alto gasto.

O grupo de alto gasto se destaca por maior renda média, maior volume de compras e menor média de filhos em casa. Já o grupo de baixo gasto tem menor renda média, menor frequência de compra e maior média de filhos.
